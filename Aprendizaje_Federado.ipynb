```python
import numpy as np
import pandas as pd
import pickle
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchtext.data.utils import get_tokenizer
from torchtext.vocab import vocab as create_vocab
from collections import Counter, OrderedDict
from torch.nn.utils.rnn import pad_sequence


# 1. Cargar metadatos y vocabulario (REEMPLAZA LA CARGA DEL DATASET COMPLETO)
label_metadata = np.load("/content/drive/My Drive/federated_text_data/label_metadata.npz", allow_pickle=True)
with open("/content/drive/My Drive/federated_text_data/vocab.pkl", "rb") as f:
    vocab = pickle.load(f)

# 2. Mapeo de etiquetas (CORRECCIÓN DEFINITIVA)
classes = label_metadata['classes'].tolist()  # Conversión segura a lista
label_to_idx = {label:i for i, label in enumerate(classes)}
```

```python
# Clase Dataset (compartida por todos los clientes)
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.texts = texts
        self.labels = labels

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        return self.texts[idx], self.labels[idx]

# 4. Función collate (sin cambios)
def collate_fn(batch):
    texts, labs = zip(*batch)
    padded = pad_sequence(texts, batch_first=True, padding_value=vocab['<pad>'])
    return padded.to(torch.long), torch.tensor(labs, dtype=torch.long)

# Arquitectura CNN
class CNNTextClassifier(nn.Module):
    def __init__(self, vocab_size, emb_dim=100, num_classes=2,
                 num_filters=100, filter_sizes=[3,4,5], dropout=0.5):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_dim, padding_idx=vocab['<pad>'])
        self.convs = nn.ModuleList([
            nn.Conv2d(1, num_filters, (fs, emb_dim)) for fs in filter_sizes
        ])
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(num_filters * len(filter_sizes), num_classes)

    def forward(self, x):
        emb = self.embedding(x).unsqueeze(1)
        conved = [torch.relu(conv(emb)).squeeze(3) for conv in self.convs]
        pooled = [torch.max(c, dim=2)[0] for c in conved]
        cat = torch.cat(pooled, dim=1)
        return self.fc(self.dropout(cat))
```

```python
import flwr as fl

class FlowerClient(fl.client.NumPyClient):
    def __init__(self, cid, simulate_failures=False):
        self.cid = cid
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        self.model = CNNTextClassifier(len(vocab), num_classes=len(classes)).to(self.device)
        self.trainloader, self.valloader = self.load_data()
        self.optimizer = optim.Adam(self.model.parameters(), lr=1e-3)
        self.criterion = nn.CrossEntropyLoss()
        self.simulate_failures = simulate_failures

    def load_data(self):
        data = np.load(f"/content/drive/My Drive/federated_text_data/client_{self.cid}.npz", allow_pickle=True)

        # Conversión robusta
        texts = [torch.tensor(seq.astype(np.int64)) for seq in data['texts']]
        labels = torch.tensor(data['labels'].astype(np.int64))

        # División 80-20
        split_idx = int(0.8 * len(texts))
        return (
            DataLoader(
                TextDataset(texts[:split_idx], labels[:split_idx]),
                batch_size=32,
                shuffle=True,
                collate_fn=collate_fn
            ),
            DataLoader(
                TextDataset(texts[split_idx:], labels[split_idx:]),
                batch_size=64,
                shuffle=False,
                collate_fn=collate_fn
            )
        )


    def get_parameters(self, config):
        return [val.cpu().numpy() for _, val in self.model.state_dict().items()]

    def fit(self, parameters, config):
        if self.simulate_failure():  # Simulación de fallo
            return self.get_parameters({}), 0, {"accuracy": 0.0, "failed": True}

        self.model.load_state_dict({k: torch.tensor(v) for k, v in zip(self.model.state_dict().keys(), parameters)})

        total_samples = 0
        correct = 0
        self.model.train()
        for _ in range(3):
            for texts, labels in self.trainloader:
                texts, labels = texts.to(self.device), labels.to(self.device)
                self.optimizer.zero_grad()
                outputs = self.model(texts)
                loss = self.criterion(outputs, labels)
                loss.backward()
                self.optimizer.step()

                _, predicted = torch.max(outputs.data, 1)
                total_samples += labels.size(0)
                correct += (predicted == labels).sum().item()

        accuracy = correct / total_samples if total_samples > 0 else 0.0
        return self.get_parameters({}), len(self.trainloader.dataset), {"accuracy": accuracy, "failed": False}

    def evaluate(self, parameters, config):
        self.model.load_state_dict({k: torch.tensor(v) for k, v in zip(self.model.state_dict().keys(), parameters)})
        self.model.eval()

        correct, total, loss = 0, 0, 0.0
        with torch.no_grad():
            for texts, labels in self.valloader:
                texts, labels = texts.to(self.device), labels.to(self.device)
                outputs = self.model(texts)
                loss += self.criterion(outputs, labels).item() * labels.size(0)
                total += labels.size(0)
                correct += (outputs.argmax(1) == labels).sum().item()

        accuracy = correct / total if total > 0 else 0.0
        return float(loss/total), total, {"accuracy": accuracy}

    def get_dense_stats(self):
        """Calcula el peso real de los parámetros para el modelo denso."""
        # Contamos todos los parámetros del modelo
        total_params = sum(p.numel() for p in self.model.parameters())

        # En el modelo denso, el 100% de los parámetros son "activos"
        active_params = total_params

        # Peso en MB (4 bytes por parámetro float32)
        mb_size = (active_params * 4) / (1024**2)
        return total_params, active_params, mb_size

    def simulate_failure(self):
        if not self.simulate_failures:
            return False
        # Simulación simple de fallo con 20% de probabilidad
        return np.random.random() < 0.2
```

```python
class CustomStrategy(fl.server.strategy.FedAvg):
    def __init__(self):
        super().__init__(evaluate_metrics_aggregation_fn=self.aggregate_metrics)
        self.best_accuracy = 0.0
        self.best_model = None
        self.history_accuracy = []
        self.history_network_mb = []

    def aggregate_fit(self, server_round, results, failures):
        # Filtrar clientes fallidos
        valid_results = [r for r in results if not r[1].metrics.get("failed", False)]
        accuracies = [res.metrics["accuracy"] for _, res in valid_results]

        if accuracies:
            best_idx = np.argmax(accuracies)
            best_accuracy = accuracies[best_idx]

            if best_accuracy > self.best_accuracy:
                self.best_accuracy = best_accuracy
                self.best_model = valid_results[best_idx][1].parameters
                print(f"✅ Nuevo mejor modelo (Accuracy: {best_accuracy:.2%})")

        return super().aggregate_fit(server_round, results, failures)

    def aggregate_metrics(self, metrics):  # Nueva función para registrar métricas
        accuracies = [m["accuracy"] for _, m in metrics]
        avg_accuracy = np.mean(accuracies) if accuracies else 0.0
        return {"accuracy": avg_accuracy}

    def aggregate_evaluate(self, server_round, results, failures):
        """Calcula Accuracy y Tráfico de Red Real para el modelo denso."""
        # 1. Accuracy promedio
        accuracies = [r.metrics["accuracy"] for _, r in results]
        avg_accuracy = np.mean(accuracies) if accuracies else 0.0
        self.history_accuracy.append(avg_accuracy)

        # 2. CÁLCULO DE RED REAL (DENSO)
        # Obtenemos un cliente para medir el peso total del modelo
        test_client = client_fn("0")
        _, _, mb_per_client = test_client.get_dense_stats()

        # Tráfico: Clientes que respondieron * peso total * 2 (ida y vuelta)
        total_mb_round = len(results) * mb_per_client * 2
        self.history_network_mb.append(total_mb_round)

        # 3. Log para consola
        print(f"--- Ronda {server_round} ---")
        print(f"📈 Accuracy: {avg_accuracy:.2%}")
        print(f"📡 Tráfico Red: {total_mb_round:.2f} MB (Modelo DENSO)")
        print("-" * 20)

        loss_aggregated, _ = super().aggregate_evaluate(server_round, results, failures)
        return loss_aggregated, {"accuracy": avg_accuracy, "network_mb": total_mb_round}

def client_fn(cid: str) -> FlowerClient:
    return FlowerClient(int(cid))  # Habilitar simulación de fallos
# Crear la estrategia y guardarla en una variable
strategy = CustomStrategy()
# Ejecutar simulación
history = fl.simulation.start_simulation(
    client_fn=client_fn,
    num_clients=5,
    config=fl.server.ServerConfig(num_rounds=10),
    strategy=strategy,
    client_resources={"num_cpus": 1, "num_gpus": 0.5 if torch.cuda.is_available() else 0},
)
# --- REPORTE TÉCNICO REAL (DENSO) ---
print("\n" + "="*50)
print("📡 MÉTRICAS DE RED: MODELO DENSO (BASELINE)".center(50))
print("="*50)

# Datos de la arquitectura
test_client = client_fn("0")
total_p, active_p, mb_round = test_client.get_dense_stats()

# Tráfico total acumulado
total_transmision_mb = sum(strategy.history_network_mb)

print(f"1. Estructura del Modelo:")
print(f"   - Parámetros Totales: {total_p:,}")
print(f"   - Parámetros Activos: {active_p:,} (100.0%)")

print(f"\n2. Carga de Comunicación (Overhead):")
print(f"   - Payload por cliente (Subida): {mb_round:.2f} MB")
print(f"   - Tráfico total en la simulación: {total_transmision_mb:.2f} MB")

print(f"\n3. Estado para el Workshop:")
print(f"   - Este es el baseline denso (Sparsity: 0%)")
print(f"   - Cada comunicación transmite el modelo completo.")
print("="*50)
```

```python
class SimpleRNN(nn.Module):
    def __init__(self, input_size, hidden_size=128, dropout=0.3, num_classes=23):
        super(SimpleRNN, self).__init__()

        # Primera capa densa con normalización y dropout
        self.fc1 = nn.Linear(input_size, hidden_size*2)
        self.bn1 = nn.BatchNorm1d(hidden_size*2)
        self.dropout1 = nn.Dropout(dropout)

        # Segunda capa densa
        self.fc2 = nn.Linear(hidden_size*2, hidden_size)
        self.bn2 = nn.BatchNorm1d(hidden_size)
        self.dropout2 = nn.Dropout(dropout)

        # Capa de salida
        self.fc3 = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        # Si los datos son 3D (como para LSTM), aplanarlos
        if len(x.shape) > 2:
            batch_size = x.size(0)
            x = x.reshape(batch_size, -1)

        # Propagación por las capas
        x = F.relu(self.bn1(self.fc1(x)))
        x = self.dropout1(x)

        x = F.relu(self.bn2(self.fc2(x)))
        x = self.dropout2(x)

        x = self.fc3(x)
        return x
```

## 2. Desenvolvendo Classe Servidor

```python
from logging import WARNING
from flwr.server.strategy.aggregate import aggregate, weighted_loss_avg
from flwr.server.strategy.strategy import Strategy

from flwr.common import (
    EvaluateIns,
    EvaluateRes,
    FitIns,
    FitRes,
    MetricsAggregationFn,
    NDArrays,
    Parameters,
    Scalar,
    ndarrays_to_parameters,
    parameters_to_ndarrays,
)

from flwr.common.logger import log
from flwr.server.client_manager import ClientManager
from flwr.server.client_proxy import ClientProxy


class AgregacaoCustomizada(fl.server.strategy.FedAvg):

    def configure_fit(
        self, server_round, parameters, client_manager):
        """Configure the next round of training."""
        config = {
           "servidor_round" : server_round
            }

        if self.on_fit_config_fn is not None:
            config = self.on_fit_config_fn(server_round)
        else:
            config = {"servidor_round": server_round}

        fit_ins = fl.common.FitIns(parameters, config)

        # Sample clients
        sample_size, min_num_clients = self.num_fit_clients(
            client_manager.num_available()
        )
        clients = client_manager.sample(
            num_clients=sample_size, min_num_clients=min_num_clients
        )

        # Return client/config pairs
        return [(client, fit_ins) for client in clients]

    def aggregate_fit(
        self,
        server_round,
        results ,
        failures ,
    ):

        """Aggregate fit results using weighted average."""
        if not results:
            return None, {}
        # Do not aggregate if there are failures and failures are not accepted
        if not self.accept_failures and failures:
            return None, {}

        # Convert results
        weights_results = [
            (parameters_to_ndarrays(fit_res.parameters), fit_res.num_examples)
            for _, fit_res in results
        ]
        parameters_aggregated = ndarrays_to_parameters(aggregate(weights_results))

        print(f"Epocha {server_round}")
        # Aggregate custom metrics if aggregation fn was provided
        metrics_aggregated = {}
        if self.fit_metrics_aggregation_fn:
            fit_metrics = [(res.num_examples, res.metrics) for _, res in results]
            metrics_aggregated = self.fit_metrics_aggregation_fn(fit_metrics)
        elif server_round == 1:  # Only log this warning once
            log(WARNING, "No fit_metrics_aggregation_fn provided")

        return parameters_aggregated, metrics_aggregated


    def aggregate_evaluate(
        self,
        server_round,
        results,
        failures,
    ):
        """Aggregate evaluation losses using weighted average."""
        if not results:
            return None, {}
        # Do not aggregate if there are failures and failures are not accepted
        if not self.accept_failures and failures:
            return None, {}

        # Aggregate loss
        loss_aggregated = weighted_loss_avg(
            [
                (evaluate_res.num_examples, evaluate_res.loss)
                for _, evaluate_res in results
            ]
        )

        # Aggregate custom metrics if aggregation fn was provided
        metrics_aggregated = {}
        if self.evaluate_metrics_aggregation_fn:
            eval_metrics = [(res.num_examples, res.metrics) for _, res in results]
            metrics_aggregated = self.evaluate_metrics_aggregation_fn(eval_metrics)
        elif server_round == 1:  # Only log this warning once
            log(WARNING, "No evaluate_metrics_aggregation_fn provided")

        return loss_aggregated, metrics_aggregated
```
